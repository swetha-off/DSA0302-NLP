```python

import re
import textwrap

hospital_records = [
    {
        "id": "PAT001",
        "type": "Discharge Summary",
        "text": """
        Patient admitted on 12/03/2025.
        Diagnosis: Femoral fracture.
        Patient underwent surgery.
        Discharged on 18/03/2025.
        Prescribed Paracetamol 500mg twice daily.
        """
    },
    {
        "id": "PAT002",
        "type": "Prescription",
        "text": """
        Patient has fractured left tibia.
        Prescribed Paracetamol 500 mg BD.
        Blood pressure 120/80 mmHg.
        Temperature 37.5 C.
        """
    },
    {
        "id": "PAT003",
        "type": "Operation Note",
        "text": """
        Patient underwent nephrectomy.
        Renal condition was monitored.
        Surgery completed successfully.
        """
    },
    {
        "id": "PAT004",
        "type": "Nursing Observation",
        "text": """
        Patient reports pain for 3 days.
        Temperature 38 C.
        Blood pressure 130/85 mmHg.
        Patient is recovering.
        """
    }
]

linguistic_levels = {
    "Phonology": {
        "meaning": "Study of speech sounds and sound patterns.",
        "example": "Different pronunciations of regional language words.",
        "requirement": "Useful for OCR and Romanised language input."
    },

    "Morphology": {
        "meaning": "Study of word formation and word structure.",
        "example": "Fracture, fractures, fractured, fracturing.",
        "requirement": "Essential for identifying related word forms."
    },

    "Syntax": {
        "meaning": "Study of grammatical sentence structure.",
        "example": "Patient was diagnosed with a fracture.",
        "requirement": "Useful for extracting relations between words."
    },

    "Semantics": {
        "meaning": "Study of meaning in language.",
        "example": "Nephrectomy is related to kidney removal.",
        "requirement": "Important for connecting clinical concepts."
    },

    "Pragmatics": {
        "meaning": "Study of meaning based on context.",
        "example": "Rule out fracture does not mean fracture confirmed.",
        "requirement": "Important for interpreting clinical statements."
    },

    "Discourse": {
        "meaning": "Study of relationships between sentences.",
        "example": "Patient underwent surgery. He recovered well.",
        "requirement": "Useful for connecting information across notes."
    }
}


def display_linguistic_levels():

    print("\n" + "=" * 65)
    print("1. LEVELS OF LINGUISTIC ANALYSIS")
    print("=" * 65)

    for level, details in linguistic_levels.items():

        print("\nLevel:", level)
        print("Meaning:", details["meaning"])
        print("Clinical Example:", details["example"])
        print("Requirement:", details["requirement"])

model_families = {
    "State Machines": {
        "task": "Validating patient identifiers and date formats.",
        "reason": "Finite-state automata recognise structured patterns.",
        "wrong_choice": "Probabilistic models are unnecessary for exact format validation. "
                        "General rule systems are less convenient for representing "
                        "state-by-state recognition."
    },

    "Formal Rule Systems": {
        "task": "Morphological analysis and word normalisation.",
        "reason": "Rules identify prefixes, suffixes, stems and irregular forms.",
        "wrong_choice": "State machines alone do not conveniently express all "
                        "morphological relationships. Probabilistic models require "
                        "training data and may produce inconsistent results."
    },

    "Probabilistic Models": {
        "task": "Handling ambiguous OCR words and misspelt drug names.",
        "reason": "Probability can rank alternative interpretations using context.",
        "wrong_choice": "A deterministic automaton cannot rank likely interpretations. "
                        "Fixed rules cannot easily handle every noisy input."
    }
}


def display_models():

    print("\n" + "=" * 65)
    print("2. THREE MODEL FAMILIES")
    print("=" * 65)

    for model, details in model_families.items():

        print("\nModel:", model)
        print("Suitable Task:", details["task"])
        print("Reason:", details["reason"])
        print("Why alternatives are unsuitable:", details["wrong_choice"])



ambiguities = {
    "Lexical Ambiguity": {
        "example": "Cold",
        "interpretation": [
            "Common cold (disease)",
            "Low temperature"
        ],
        "problem": "The same word has multiple meanings."
    },

    "Morphological Ambiguity": {
        "example": "Fractured",
        "interpretation": [
            "Fracture + ed",
            "Past-tense or participial form"
        ],
        "problem": "Different word forms must be mapped to a canonical stem."
    },

    "Segmentation Ambiguity": {
        "example": "500mgBD",
        "interpretation": [
            "500 mg BD",
            "Drug strength followed by dosage frequency"
        ],
        "problem": "OCR or missing spaces can merge separate tokens."
    },

    "Abbreviation Ambiguity": {
        "example": "OD",
        "interpretation": [
            "Once daily",
            "Right eye (oculus dexter)",
            "Doctor"
        ],
        "problem": "An abbreviation may have different meanings in different contexts."
    }
}


def display_ambiguities():

    print("\n" + "=" * 65)
    print("3. AMBIGUITIES IN NATURAL LANGUAGE")
    print("=" * 65)

    for ambiguity, details in ambiguities.items():

        print("\nType:", ambiguity)
        print("Example:", details["example"])

        print("Possible Interpretations:")

        for interpretation in details["interpretation"]:
            print(" -", interpretation)

        print("Problem:", details["problem"])



search_techniques = {
    "State-space Search": {
        "application": "Traversing finite-state automata.",
        "time": "O(n) for a deterministic finite automaton.",
        "memory": "O(number of automaton states).",
        "data": "No large annotated dataset required.",
        "expert_effort": "Designing states and transitions."
    },

    "Backtracking": {
        "application": "Trying alternative matching paths in a nondeterministic automaton.",
        "time": "Can be exponential in the worst case for naive matching.",
        "memory": "Depends on the number of active paths and stored states.",
        "data": "No training data required.",
        "expert_effort": "Designing and testing matching rules."
    },

    "Probabilistic Search": {
        "application": "Selecting the most likely interpretation of noisy clinical text.",
        "time": "Depends on model size and number of candidate interpretations.",
        "memory": "Model parameters and candidate storage.",
        "data": "Requires labelled examples for training or evaluation.",
        "expert_effort": "Clinical annotation and model evaluation."
    }
}


def display_search_resources():

    print("\n" + "=" * 65)
    print("4. SEARCH TECHNIQUES AND RESOURCE REQUIREMENTS")
    print("=" * 65)

    for technique, details in search_techniques.items():

        print("\nTechnique:", technique)

        for key, value in details.items():
            print(key.replace("_", " ").title() + ":", value)



pipeline = [
    ("INPUT", "Hospital clinical records"),
    ("STAGE 1", "Text acquisition and OCR"),
    ("STAGE 2", "Text cleaning and normalisation"),
    ("STAGE 3", "Tokenisation using regular expressions"),
    ("STAGE 4", "Structured entity recognition using regex"),
    ("STAGE 5", "Entity validation using finite-state automata"),
    ("STAGE 6", "Morphological parsing using lexicon and affix rules"),
    ("STAGE 7", "Stemming and canonical form generation"),
    ("STAGE 8", "Indexing using canonical forms"),
    ("STAGE 9", "Clinical query processing and retrieval"),
    ("OUTPUT", "Relevant hospital records")
]

formalisms = [
    "Input files and text extraction",
    "OCR and character processing",
    "Regular expressions and normalisation rules",
    "Regular expressions",
    "Finite-state automata (DFA/NFA)",
    "Lexicon and morphological rules",
    "Rule-based stemming",
    "Inverted index and data structures",
    "Token matching and search algorithms",
    "Ranked or matching clinical records",
    "Output results formatting"
]


def display_pipeline():

    print("\n" + "=" * 65)
    print("5. HOSPITAL NLP PIPELINE")
    print("=" * 65)

    for index, (stage, description) in enumerate(pipeline):

        print("\n" + stage)
        print("   " + description)
        print("   Formalism:", formalisms[index])

        if index < len(pipeline) - 1:
            print("       |")
            print("       v")



def normalise_text(text):

    text = text.lower()

    # Standardise common clinical abbreviations.
    abbreviations = {
        "bd": "twice daily",
        "bid": "twice daily",
        "tds": "three times daily",
        "tid": "three times daily",
        "od": "once daily"
    }

    for abbreviation, meaning in abbreviations.items():

        pattern = r"\b" + re.escape(abbreviation) + r"\b"

        text = re.sub(pattern, meaning, text)

    # Remove punctuation.
    text = re.sub(r"[^\w\s./-]", " ", text)

    # Remove extra spaces.
    text = re.sub(r"\s+", " ", text).strip()

    return text



def stem_word(word):

    irregular_forms = {
        "fractures": "fracture",
        "fractured": "fracture",
        "fracturing": "fracture",
        "nephrectomies": "nephrectomy",
        "children": "child",
        "men": "man",
        "women": "woman"
    }

    if word in irregular_forms:
        return irregular_forms[word]

    suffixes = [
        "ization",
        "ational",
        "fulness",
        "ments",
        "ingly",
        "ation",
        "ities",
        "ingly",
        "ing",
        "ed",
        "es",
        "s"
    ]

    for suffix in suffixes:

        if word.endswith(suffix) and len(word) > len(suffix) + 2:

            return word[:-len(suffix)]

    return word



def search_records(query):

    print("\nSearching for:", query)

    query = normalise_text(query)

    query_words = query.split()

    query_stems = [stem_word(word) for word in query_words]

    results = []

    for record in hospital_records:

        normalised_record = normalise_text(record["text"])

        record_words = normalised_record.split()

        record_stems = [
            stem_word(word) for word in record_words
        ]

        # Check whether any query stem occurs in the record.
        if any(stem in record_stems for stem in query_stems):

            results.append(record)

    if results:

        print("\nMatching Records:")

        for record in results:

            print("\nPatient ID:", record["id"])
            print("Record Type:", record["type"])
            print("Clinical Text:")

            print(textwrap.fill(record["text"].strip(), width=70))

    else:

        print("\nNo matching records found.")

def main():

    print("\n")
    print("*" * 65)
    print("HOSPITAL TEXT NORMALISATION AND SEARCH SYSTEM")
    print("TASK 1: FRAME THE PROBLEM")
    print("*" * 65)

    display_linguistic_levels()

    display_models()

    display_ambiguities()

    display_search_resources()

    display_pipeline()

    # Demonstrate morphological normalisation.
    print("\n" + "=" * 65)
    print("6. MORPHOLOGICAL NORMALISATION")
    print("=" * 65)

    test_words = [
        "fracture",
        "fractures",
        "fractured",
        "fracturing",
        "nephrectomies",
        "children"
    ]

    for word in test_words:

        print(word, "---->", stem_word(word))

    # Demonstrate hospital record retrieval.
    print("\n" + "=" * 65)
    print("7. HOSPITAL RECORD SEARCH")
    print("=" * 65)

    search_records("fracture")

    search_records("fractured")

    search_records("nephrectomy")


if __name__ == "__main__":
    main()
```

```python

import re
from datetime import datetime

# 1. SINGLE REGEX TOKENIZER

TOKEN_PATTERN = re.compile(
    r"""
    \d{4}[-/]\d{2}[-/]\d{2}                 # YYYY-MM-DD
    |
    \d{1,2}[-/]\d{1,2}[-/]\d{2,4}           # DD/MM/YYYY
    |
    \d+(?:\.\d+)?(?:mg|mcg|µg|ml|mL|g|kg|mmHg|bpm|°C|°F)\b
    |
    (?:[A-Za-z]\.){2,}                       # e.g. U.S.
    |
    [A-Za-z]{1,3}\.                         # Dr. etc.
    |
    \d+(?:\.\d+)?                            # Decimal/integer
    |
    [A-Za-z]+(?:-[A-Za-z0-9]+)+             # Hyphenated terms
    |
    [A-Za-z]+                                # Words
    |
    [^\w\s]                                 # Punctuation
    """,
    re.VERBOSE | re.IGNORECASE
)


def tokenize(text):
    return TOKEN_PATTERN.findall(text)


print("1. REGEX TOKENISATION")

clinical_text = (
    "Patient has 37.5°C fever and takes 500mg "
    "medicine. Follow-up is required."
)

print("Original Text:", clinical_text)
print("Tokens:", tokenize(clinical_text))



print("\n2. MAXIMAL-MUNCH VS WHITESPACE SPLIT")

examples = [
    "Patient has 37.5°C fever.",
    "Take 500mg paracetamol twice-daily.",
    "Dr. Kumar recorded 120/80 mmHg."
]

for sentence in examples:
    print("\nOriginal:", sentence)
    print("Whitespace Split:", sentence.split())
    print("Regex Tokens:", tokenize(sentence))



abbreviations = {
    "BP": "blood pressure",
    "HR": "heart rate",
    "RR": "respiratory rate",
    "SpO2": "oxygen saturation",
    "SOB": "shortness of breath",
    "COPD": "chronic obstructive pulmonary disease",
    "CHF": "congestive heart failure",
    "DM": "diabetes mellitus",
    "HTN": "hypertension",
    "CAD": "coronary artery disease",
    "MI": "myocardial infarction",
    "CVA": "cerebrovascular accident",
    "UTI": "urinary tract infection",
    "URI": "upper respiratory infection",
    "GI": "gastrointestinal",
    "ECG": "electrocardiogram",
    "EKG": "electrocardiogram",
    "CBC": "complete blood count",
    "WBC": "white blood cell count",
    "RBC": "red blood cell count",
    "Hb": "hemoglobin",
    "Hgb": "hemoglobin",
    "Hct": "hematocrit",
    "Na": "sodium",
    "K": "potassium",
    "Cr": "creatinine",
    "BUN": "blood urea nitrogen",
    "PRN": "as needed",
    "BID": "twice daily",
    "TID": "three times daily",
    "QID": "four times daily",
    "PO": "by mouth",
    "IV": "intravenous",
    "IM": "intramuscular",
    "SC": "subcutaneous",
    "NPO": "nothing by mouth",
    "ICU": "intensive care unit",
    "ER": "emergency room",
    "OPD": "outpatient department",
    "STAT": "immediately",
    "y/o": "years old",
    "c/o": "complains of"
}



def expand_abbreviations(text):

    keys = sorted(abbreviations.keys(), key=len, reverse=True)

    for short_form in keys:
        pattern = r"(?<!\w)" + re.escape(short_form) + r"(?!\w)"

        text = re.sub(
            pattern,
            abbreviations[short_form],
            text,
            flags=re.IGNORECASE
        )

    return text


def normalize_dates(text):

    def convert(match):
        value = match.group()

        for fmt in ("%d/%m/%Y", "%d-%m-%Y"):
            try:
                date = datetime.strptime(value, fmt)
                return date.strftime("%Y-%m-%d")
            except ValueError:
                pass

        return value

    text = re.sub(
        r"\b\d{1,2}[/\-]\d{1,2}[/\-]\d{4}\b",
        convert,
        text
    )

    return text


def normalize_units(text):

    unit_map = {
        "milligrams": "mg",
        "milligram": "mg",
        "mg": "mg",
        "micrograms": "mcg",
        "microgram": "mcg",
        "mcg": "mcg",
        "µg": "mcg",
        "grams": "g",
        "gram": "g",
        "g": "g",
        "milliliters": "mL",
        "milliliter": "mL",
        "ml": "mL",
        "liters": "L",
        "liter": "L",
        "kg": "kg",
        "mmhg": "mmHg",
        "bpm": "bpm",
        "c": "°C",
        "°c": "°C",
        "f": "°F",
        "°f": "°F"
    }

    unit_pattern = (
        r"(?<=\d)\s*"
        r"(milligrams|milligram|mg|micrograms|microgram|mcg|µg|"
        r"milliliters|milliliter|ml|liters|liter|kg|grams|gram|g|"
        r"mmhg|bpm|°c|°f|c|f)\b"
    )

    def replace_unit(match):
        unit = match.group().strip().lower()
        return " " + unit_map[unit]

    return re.sub(
        unit_pattern,
        replace_unit,
        text,
        flags=re.IGNORECASE
    )


def normalize_text(text):

    text = text.casefold()

    text = expand_abbreviations(text)

    text = normalize_dates(text)

    text = normalize_units(text)

    text = re.sub(r"\s+", " ", text).strip()

    return text



print("\n3. CLINICAL TEXT NORMALISATION")

record = (
    "PATIENT P101: 45 y/o male, c/o SOB. "
    "BP 120/80 mmHg, HR 90 bpm, temperature 37.5 C. "
    "SpO2 98%. Prescribed 500mg medication BID. "
    "Diagnosis: HTN and DM. Admission date: 06/11/2026."
)

normalized_record = normalize_text(record)

print("\nBEFORE NORMALISATION:")
print(record)

print("\nAFTER NORMALISATION:")
print(normalized_record)


print("\n4. FINAL NORMALISED TOKEN STREAM")

tokens = tokenize(normalized_record)

for i, token in enumerate(tokens, start=1):
    print(i, ":", token)



print("\n5. IRREVERSIBLE NORMALISATION")

losses = [
    "Case folding: BP and bp become identical.",
    "Abbreviation expansion: Original short forms are lost.",
    "Date conversion: Original date format is lost.",
    "Whitespace removal: Original spacing is lost."
]

for item in losses:
    print("-", item)



print("\n6. LIMITATIONS OF REGEX")

nested_note = "Patient history (diabetes (Type 2)) recorded."

print("Clinical Note:", nested_note)

print(
    "A simple regex cannot reliably identify "
    "arbitrarily nested balanced parentheses."
)

print("\nTASK 3 COMPLETED SUCCESSFULLY")
```

```python

# ==============================================
# TASK 4: FINITE STATE AUTOMATA
# ==============================================

from collections import defaultdict, deque

# ----------------------------------------------
# 1. FORMAL DEFINITION OF FINITE AUTOMATON
# ----------------------------------------------

print("1. FORMAL DEFINITION OF FINITE AUTOMATON")

print("""
A Finite Automaton is represented by:
M = (Q, Sigma, delta, q0, F)

Q     = Set of states
Sigma = Input alphabet
delta = Transition function
q0    = Initial state
F     = Set of final states
""")


# ----------------------------------------------
# 2. DFA DESIGN FOR FOUR CLINICAL VALIDATORS
# ----------------------------------------------

# Each DFA uses character classes.
# D = digit, P = literal P, Y = literal Y,
# M = literal M, A = literal A, G = literal G.

automata = {

    "Patient ID": {
        "start": "q0",
        "final": {"q4"},
        "transitions": {
            ("q0", "P"): "q1",
            ("q1", "D"): "q2",
            ("q2", "D"): "q3",
            ("q3", "D"): "q4"
        }
    },

    "Age (number + Y)": {
        "start": "q0",
        "final": {"q2", "q3", "q4"},
        "transitions": {
            ("q0", "D"): "q2",
            ("q2", "D"): "q3",
            ("q3", "D"): "q4"
        }
    },

    "Dosage (number + MG)": {
        "start": "q0",
        "final": {"q3"},
        "transitions": {
            ("q0", "D"): "q1",
            ("q1", "D"): "q1",
            ("q1", "M"): "q2",
            ("q2", "G"): "q3"
        }
    },

    "Clinical code (P + 3 digits + A)": {
        "start": "q0",
        "final": {"q5"},
        "transitions": {
            ("q0", "P"): "q1",
            ("q1", "D"): "q2",
            ("q2", "D"): "q3",
            ("q3", "D"): "q4",
            ("q4", "A"): "q5"
        }
    }
}


def get_symbol(ch):
    if ch.isdigit():
        return "D"
    return ch.upper()


def validate_dfa(text, dfa):
    state = dfa["start"]

    for ch in text:
        symbol = get_symbol(ch)
        key = (state, symbol)

        if key not in dfa["transitions"]:
            return False

        state = dfa["transitions"][key]

    return state in dfa["final"]


print("\n2. DFA TRANSITION TABLES")

for name, dfa in automata.items():

    print("\nDFA:", name)
    print("Initial State:", dfa["start"])
    print("Final States:", dfa["final"])

    print("Transition Table:")
    print("Current State | Input | Next State")

    for (state, symbol), next_state in dfa["transitions"].items():
        print(state, "|", symbol, "|", next_state)

    print("Transition Diagram:")

    for (state, symbol), next_state in dfa["transitions"].items():
        print(state, "--", symbol, "-->", next_state)


# Test the four DFA validators

print("\nDFA VALIDATION")

tests = {
    "Patient ID": ["P123", "P12", "P1234"],
    "Age (number + Y)": ["45", "120", "1234"],
    "Dosage (number + MG)": ["500MG", "250MG", "50ML"],
    "Clinical code (P + 3 digits + A)": ["P123A", "P12A", "P123B"]
}

for name, values in tests.items():

    print("\n", name)

    for value in values:
        result = validate_dfa(value, automata[name])
        print(value, ":", "ACCEPTED" if result else "REJECTED")


# ----------------------------------------------
# 3. NFA WITH EPSILON TRANSITIONS
# ----------------------------------------------

print("\n3. NFA WITH EPSILON TRANSITIONS")

# Language: P123 or P123A
# The final A is optional.

nfa = {
    0: {"P": {1}},
    1: {"D": {2}},
    2: {"D": {3}},
    3: {"D": {4}},
    4: {"eps": {5, 6}},
    5: {},
    6: {"A": {5}}
}

nfa_start = 0
nfa_final = {4, 5}


def epsilon_closure(states):
    closure = set(states)
    stack = list(states)

    while stack:
        state = stack.pop()

        for next_state in nfa.get(state, {}).get("eps", set()):
            if next_state not in closure:
                closure.add(next_state)
                stack.append(next_state)

    return closure


def nfa_move(states, symbol):
    result = set()

    for state in states:
        result.update(
            nfa.get(state, {}).get(symbol, set())
        )

    return result


def trace_nfa(string):

    current = epsilon_closure({nfa_start})

    print("\nInput:", string)
    print("Initial epsilon closure:", current)

    for ch in string:
        symbol = "D" if ch.isdigit() else ch.upper()

        moved = nfa_move(current, symbol)
        current = epsilon_closure(moved)

        print("Input:", ch, "Live States:", current)

    accepted = bool(current.intersection(nfa_final))

    print("Result:", "ACCEPTED" if accepted else "REJECTED")


trace_nfa("P123")
trace_nfa("P123A")
trace_nfa("P12A")


# ----------------------------------------------
# 4. NFA TO DFA USING SUBSET CONSTRUCTION
# ----------------------------------------------

print("\n4. NFA TO DFA SUBSET CONSTRUCTION")

alphabet = ["P", "D", "A"]

start_set = frozenset(epsilon_closure({nfa_start}))

dfa_states = [start_set]
dfa_transitions = {}
queue = deque([start_set])

while queue:

    current = queue.popleft()

    for symbol in alphabet:

        next_set = epsilon_closure(
            nfa_move(current, symbol)
        )

        next_set = frozenset(next_set)

        dfa_transitions[(current, symbol)] = next_set

        if next_set not in dfa_states:
            dfa_states.append(next_set)
            queue.append(next_set)


dfa_final = [
    state for state in dfa_states
    if state.intersection(nfa_final)
]

print("\nSubset Construction Table")
print("DFA State | NFA States | P | D | A | Final")

for index, state in enumerate(dfa_states):

    outputs = []

    for symbol in alphabet:
        target = dfa_transitions[(state, symbol)]
        target_index = dfa_states.index(target)
        outputs.append("S" + str(target_index))

    print(
        "S" + str(index),
        "|", set(state),
        "|", outputs[0],
        "|", outputs[1],
        "|", outputs[2],
        "|", state in dfa_final
    )


# ----------------------------------------------
# 5. DFA MINIMISATION
# ----------------------------------------------

print("\n5. DFA MINIMISATION")

# Partition refinement algorithm.

all_states = set(dfa_states)

final_partition = set(dfa_final)
nonfinal_partition = all_states - final_partition

partitions = [
    part for part in [final_partition, nonfinal_partition]
    if part
]


def find_partition(state, groups):
    for i, group in enumerate(groups):
        if state in group:
            return i
    return -1


changed = True

while changed:

    changed = False
    new_partitions = []

    for group in partitions:

        buckets = {}

        for state in group:

            signature = tuple(
                find_partition(
                    dfa_transitions[(state, symbol)],
                    partitions
                )
                for symbol in alphabet
            )

            buckets.setdefault(signature, set()).add(state)

        new_partitions.extend(buckets.values())

        if len(buckets) > 1:
            changed = True

    partitions = new_partitions


print("DFA States Before Minimisation:", len(dfa_states))
print("DFA States After Minimisation:", len(partitions))

print("\nMinimised State Groups:")

for i, group in enumerate(partitions):
    print("M" + str(i), "=", [set(s) for s in group])


# ----------------------------------------------
# 6. THOMPSON-STYLE CONSTRUCTION
# ----------------------------------------------

print("\n6. THOMPSON CONSTRUCTION")

print("Regular Expression: P[0-9]{3}")

thompson = {
    "q0": {"P": "q1"},
    "q1": {"D": "q2"},
    "q2": {"D": "q3"},
    "q3": {"D": "q4"}
}

for state, transitions in thompson.items():
    for symbol, target in transitions.items():
        print(state, "--", symbol, "-->", target)

print("Initial State: q0")
print("Final State: q4")


# ----------------------------------------------
# 7. REGEX DERIVED FROM AUTOMATON
# ----------------------------------------------

print("\n7. REGEX AND AUTOMATON EQUIVALENCE")

derived_regex = r"^P[0-9]{3}$"

equivalence_tests = [
    "P123",
    "P456",
    "P12",
    "P1234",
    "A123"
]

for value in equivalence_tests:

    regex_result = bool(
        re.fullmatch(derived_regex, value)
    )

    automaton_result = validate_dfa(
        value, automata["Patient ID"]
    )

    print(
        value,
        "| Regex:", regex_result,
        "| DFA:", automaton_result
    )


# ----------------------------------------------
# 8. PUMPING LEMMA
# ----------------------------------------------

print("\n8. PUMPING LEMMA")

print("""
Pumping Lemma:

For every regular language L, there exists a pumping
length p such that any string s in L with length >= p
can be written as s = xyz, satisfying:

1. |xy| <= p
2. |y| > 0
3. xy^i z belongs to L for every i >= 0.

Consider the language:

L = { ( ^ n ) ^ n : n >= 0 }

Assume L is regular and let p be its pumping length.

Choose s = '(' repeated p times followed by
')' repeated p times.

Since |xy| <= p, y contains only opening brackets.

Removing y (i = 0) produces fewer opening brackets
than closing brackets.

Therefore, the resulting string is not in L.

This is a contradiction. Hence L is not regular.
""")

print("\nTASK 4 COMPLETED SUCCESSFULLY")
```

```python

# ==============================================
# TASK 5: INFLECTIONAL MORPHOLOGY
# ==============================================

import re

# ----------------------------------------------
# 1. MORPHOLOGICAL DEFINITIONS
# ----------------------------------------------

print("1. MORPHOLOGICAL DEFINITIONS")

definitions = {
    "Morpheme": "Smallest meaningful unit of a word.",
    "Stem": "The basic form to which an affix is added.",
    "Affix": "A prefix or suffix attached to a stem.",
    "Inflection": "Changes grammatical form without changing word category.",
    "Paradigm": "A complete set of grammatical forms of a word."
}

for term, meaning in definitions.items():
    print(term, ":", meaning)


print("\nINFLECTION VS DERIVATION")

print("""
Inflection:
- Preserves the grammatical category.
- Expresses number, tense or comparison.
- Example: patient -> patients

Derivation:
- Creates a new word or changes its meaning/category.
- Example: treat -> treatment
""")


# ----------------------------------------------
# 2. INFLECTIONAL PARADIGMS
# ----------------------------------------------

print("\n2. CLINICAL WORD PARADIGMS")

paradigms = {
    "Patient": {
        "noun": ["patient", "patients"]
    },

    "Treat": {
        "verb": [
            "treat",
            "treats",
            "treated",
            "treating"
        ]
    },

    "Monitor": {
        "verb": [
            "monitor",
            "monitors",
            "monitored",
            "monitoring"
        ]
    },

    "Severe": {
        "adjective": [
            "severe",
            "severer",
            "severest"
        ]
    },

    "High": {
        "adjective": [
            "high",
            "higher",
            "highest"
        ]
    }
}

for word, categories in paradigms.items():
    print("\nLexeme:", word)

    for category, forms in categories.items():
        print(category, ":", ", ".join(forms))


# ----------------------------------------------
# 3. ENGLISH INFLECTIONAL SPELLING RULES
# ----------------------------------------------

print("\n3. ORTHOGRAPHIC INFLECTION RULES")


def is_consonant_y(word):
    return (
        len(word) > 1
        and word.endswith("y")
        and word[-2].lower() not in "aeiou"
    )


def plural(noun):
    if noun.endswith(("s", "x", "z", "ch", "sh")):
        return noun + "es"

    if is_consonant_y(noun):
        return noun[:-1] + "ies"

    return noun + "s"


def past_tense(verb):

    if verb.endswith("e"):
        return verb + "d"

    if is_consonant_y(verb):
        return verb[:-1] + "ied"

    if (
        len(verb) >= 3
        and verb[-1] not in "aeiouwxy"
        and verb[-2] in "aeiou"
        and verb[-3] not in "aeiou"
    ):
        return verb + verb[-1] + "ed"

    return verb + "ed"


def present_participle(verb):

    if verb.endswith("ie"):
        return verb[:-2] + "ying"

    if verb.endswith("e") and not verb.endswith("ee"):
        return verb[:-1] + "ing"

    if (
        len(verb) >= 3
        and verb[-1] not in "aeiouwxy"
        and verb[-2] in "aeiou"
        and verb[-3] not in "aeiou"
    ):
        return verb + verb[-1] + "ing"

    return verb + "ing"


def third_person(verb):

    if is_consonant_y(verb):
        return verb[:-1] + "ies"

    if verb.endswith(("s", "x", "z", "ch", "sh", "o")):
        return verb + "es"

    return verb + "s"


def comparative(adjective):

    if is_consonant_y(adjective):
        return adjective[:-1] + "ier"

    if adjective.endswith("e"):
        return adjective + "r"

    return adjective + "er"


def superlative(adjective):

    if is_consonant_y(adjective):
        return adjective[:-1] + "iest"

    if adjective.endswith("e"):
        return adjective + "st"

    return adjective + "est"


# Demonstrate the four spelling rules.

print("\nRULE 1: E-DELETION")
print("Treat ->", present_participle("treat"))
print("Counter-example: see ->", present_participle("see"))

print("\nRULE 2: CONSONANT DOUBLING")
print("Stop ->", past_tense("stop"))
print("Counter-example: visit ->", past_tense("visit"))

print("\nRULE 3: Y-TO-I")
print("Study ->", plural("study"))
print("Counter-example: boy ->", plural("boy"))

print("\nRULE 4: SIBILANT -ES")
print("Class ->", plural("class"))
print("Counter-example: gas ->", plural("gas"))


# ----------------------------------------------
# 4. IRREGULAR INFLECTIONAL FORMS
# ----------------------------------------------

print("\n4. IRREGULAR CLINICAL WORD FORMS")

irregulars = {
    "be": ["am", "is", "are", "was", "were", "been"],
    "have": ["has", "had"],
    "go": ["goes", "went", "gone"],
    "do": ["does", "did", "done"],
    "diagnosis": ["diagnoses"],
    "analysis": ["analyses"],
    "child": ["children"],
    "foot": ["feet"],
    "good": ["better", "best"]
}

for lemma, forms in irregulars.items():
    print(lemma, "->", ", ".join(forms))

print("""
Irregular forms cannot always be generated using
standard suffix rules.

Therefore, they must be stored in a lexicon
with their corresponding base forms.
""")


# ----------------------------------------------
# 5. CLINICAL CORPUS
# ----------------------------------------------

print("\n5. CLINICAL CORPUS")

corpus = [
    "The patient visits the hospital.",
    "The patients visited the hospital.",
    "The nurse treats the patient.",
    "The nurse treated the patients.",
    "The doctor is monitoring the patient.",
    "The doctor monitored the patients.",
    "The patient has severe pain.",
    "The patients have higher blood pressure.",
    "The diagnosis was recorded.",
    "The diagnoses were confirmed."
]

for sentence in corpus:
    print(sentence)


# ----------------------------------------------
# 6. BUILD WORD-TO-LEMMA DICTIONARY
# ----------------------------------------------

print("\n6. WORD LEMMATIZATION")

lemma_dictionary = {}

# Add regular inflectional forms.

for word in ["patient", "treat", "monitor"]:

    if word == "patient":
        forms = [
            word,
            plural(word)
        ]

    else:
        forms = [
            word,
            third_person(word),
            past_tense(word),
            present_participle(word)
        ]

    for form in forms:
        lemma_dictionary[form] = word


for word in ["severe", "high"]:

    forms = [
        word,
        comparative(word),
        superlative(word)
    ]

    for form in forms:
        lemma_dictionary[form] = word


# Add irregular forms.

for lemma, forms in irregulars.items():
    lemma_dictionary[lemma] = lemma

    for form in forms:
        lemma_dictionary[form] = lemma


def lemmatize(word):

    word = word.lower()

    return lemma_dictionary.get(word, word)


test_words = [
    "patient",
    "patients",
    "treated",
    "treating",
    "monitors",
    "monitored",
    "diagnosis",
    "diagnoses",
    "higher",
    "highest"
]

for word in test_words:
    print(word, "->", lemmatize(word))


# ----------------------------------------------
# 7. TYPE-TO-LEMMA RATIO
# ----------------------------------------------

print("\n7. TYPE-TO-LEMMA RATIO")

all_tokens = []

for sentence in corpus:
    words = re.findall(r"\b[a-zA-Z]+\b", sentence.lower())
    all_tokens.extend(words)

unique_types = set(all_tokens)

unique_lemmas = {
    lemmatize(word) for word in unique_types
}

type_count = len(unique_types)
lemma_count = len(unique_lemmas)

ratio = type_count / lemma_count if lemma_count else 0

print("Total Tokens:", len(all_tokens))
print("Unique Word Types:", type_count)
print("Unique Lemmas:", lemma_count)
print("Type-to-Lemma Ratio:", round(ratio, 3))


# ----------------------------------------------
# 8. INFORMATION RETRIEVAL EFFECT
# ----------------------------------------------

print("\n8. INFLECTION AND RETRIEVAL")

documents = [
    "The patient visited the hospital.",
    "The nurse treated the patients.",
    "The doctor monitored the patient.",
    "The diagnosis was confirmed."
]

queries = [
    "patients",
    "treating",
    "monitoring",
    "diagnoses",
    "hospital"
]

exact_failures = 0
inflection_recovered = 0
remaining_failures = 0

for query in queries:

    exact_match = any(
        query in re.findall(r"\b[a-zA-Z]+\b", doc.lower())
        for doc in documents
    )

    normalized_query = lemmatize(query)

    normalized_match = any(
        normalized_query in [
            lemmatize(word)
            for word in re.findall(r"\b[a-zA-Z]+\b", doc.lower())
        ]
        for doc in documents
    )

    print("\nQuery:", query)
    print("Exact Match:", exact_match)
    print("After Lemmatization:", normalized_match)

    if not exact_match:
        exact_failures += 1

        if normalized_match:
            inflection_recovered += 1
        else:
            remaining_failures += 1


proportion = (
    inflection_recovered / exact_failures
    if exact_failures else 0
)

print("\nRetrieval Summary")
print("Total Queries:", len(queries))
print("Exact Match Failures:", exact_failures)
print("Failures Recovered by Inflection:", inflection_recovered)
print("Remaining Failures:", remaining_failures)
print("Inflection Recovery Proportion:",
      round(proportion * 100, 2), "%")


# ----------------------------------------------
# 9. FINAL RESULT
# ----------------------------------------------

print("\nTASK 5 COMPLETED SUCCESSFULLY")
TASK 6
```

```python

# ==============================================
# TASK 7: FINITE-STATE MORPHOLOGICAL PARSING
# ==============================================

from collections import defaultdict

# ----------------------------------------------
# 1. FINITE STATE TRANSDUCER DEFINITION
# ----------------------------------------------

print("1. FINITE STATE TRANSDUCER")

print("""
A Finite State Transducer (FST) is a two-tape
automaton.

Lexical tape  : Contains stem and grammatical features.
Surface tape  : Contains the actual written word.

Example:
fracture + N + PL <-> fractures

The same transducer supports parsing and generation.
""")


# ----------------------------------------------
# 2. LEXICON AND CONTINUATION CLASSES
# ----------------------------------------------

print("\n2. LEXICON AND CONTINUATION CLASSES")

lexicon = {
    "NOUN": {
        "patient": ["PL"],
        "fracture": ["PL"],
        "infection": ["PL"],
        "diagnosis": ["PL"],
        "treatment": ["PL"],
        "hospital": ["PL"]
    },

    "VERB": {
        "treat": ["3SG", "PAST", "GER"],
        "monitor": ["3SG", "PAST", "GER"],
        "fracture": ["3SG", "PAST", "GER"],
        "diagnose": ["3SG", "PAST", "GER"]
    },

    "ADJECTIVE": {
        "severe": ["COMP", "SUP"],
        "high": ["COMP", "SUP"],
        "clinical": ["COMP", "SUP"]
    }
}


# ----------------------------------------------
# 3. MORPHOLOGICAL GENERATION RULES
# ----------------------------------------------

def generate_noun(stem, feature):

    if feature == "PL":

        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"

        if stem.endswith("y") and len(stem) > 1:
            if stem[-2] not in "aeiou":
                return stem[:-1] + "ies"

        return stem + "s"

    return None


def generate_verb(stem, feature):

    if feature == "3SG":

        if stem.endswith("y") and stem[-2] not in "aeiou":
            return stem[:-1] + "ies"

        if stem.endswith(("s", "x", "z", "ch", "sh", "o")):
            return stem + "es"

        return stem + "s"

    if feature == "PAST":

        if stem.endswith("e"):
            return stem + "d"

        if stem.endswith("y") and stem[-2] not in "aeiou":
            return stem[:-1] + "ied"

        if (
            len(stem) >= 3
            and stem[-1] not in "aeiouwxy"
            and stem[-2] in "aeiou"
            and stem[-3] not in "aeiou"
        ):
            return stem + stem[-1] + "ed"

        return stem + "ed"

    if feature == "GER":

        if stem.endswith("ie"):
            return stem[:-2] + "ying"

        if stem.endswith("e") and not stem.endswith("ee"):
            return stem[:-1] + "ing"

        return stem + "ing"

    return None


def generate_adjective(stem, feature):

    if feature == "COMP":

        if stem == "good":
            return "better"

        if stem.endswith("e"):
            return stem + "r"

        return stem + "er"

    if feature == "SUP":

        if stem == "good":
            return "best"

        if stem.endswith("e"):
            return stem + "st"

        return stem + "est"

    return None


def generate_surface(stem, category, feature):

    if category == "NOUN":
        return generate_noun(stem, feature)

    if category == "VERB":
        return generate_verb(stem, feature)

    if category == "ADJECTIVE":
        return generate_adjective(stem, feature)

    return None


# ----------------------------------------------
# 4. BUILD FST RELATION
# ----------------------------------------------

print("\n3. BUILDING FST RELATION")

fst = defaultdict(list)

for category, entries in lexicon.items():

    for stem, features in entries.items():

        # Include the base form.
        fst[stem].append(
            (stem, category, "BASE")
        )

        for feature in features:

            surface = generate_surface(
                stem, category, feature
            )

            if surface:
                lexical_form = stem + "+" + category[0] + "+" + feature

                fst[surface].append(
                    (stem, category, feature)
                )

                print(
                    lexical_form,
                    "<->",
                    surface
                )


# ----------------------------------------------
# 5. MORPHOLOGICAL PARSING
# ----------------------------------------------

print("\n4. MORPHOLOGICAL PARSING")


def parse_word(surface):

    return fst.get(surface.lower(), [])


def show_trace(surface, analysis):

    stem, category, feature = analysis

    print("\nSurface Form:", surface)
    print("State q0: Start")
    print("State q1: Lexicon stem recognized")
    print("State q2: Category recognized")

    if feature == "BASE":
        print("State q3: Base form accepted")
    else:
        print("State q3: Inflectional feature recognized")

    print("State q4: Final state")

    print("Lexical Tape:",
          stem + "+" + category + "+" + feature)

    print("Surface Tape:", surface)

    print("Analysis:",
          stem + "+" + category + "+" + feature)


test_words = [
    "fractures",
    "patients",
    "treated",
    "monitoring",
    "severe",
    "diagnoses",
    "diagnosed"
]

for word in test_words:

    analyses = parse_word(word)

    if analyses:

        for analysis in analyses:
            show_trace(word, analysis)

    else:
        print("\n", word, "-> No analysis found")


# ----------------------------------------------
# 6. GENERATION DIRECTION
# ----------------------------------------------

print("\n5. MORPHOLOGICAL GENERATION")

generation_inputs = [
    ("fracture", "NOUN", "PL"),
    ("patient", "NOUN", "PL"),
    ("treat", "VERB", "PAST"),
    ("monitor", "VERB", "GER"),
    ("severe", "ADJECTIVE", "COMP"),
    ("diagnose", "VERB", "PAST")
]

for stem, category, feature in generation_inputs:

    surface = generate_surface(
        stem, category, feature
    )

    print(
        stem + "+" + category + "+" + feature,
        "->",
        surface
    )


# ----------------------------------------------
# 7. AMBIGUITY HANDLING
# ----------------------------------------------

print("\n6. AMBIGUOUS MORPHOLOGICAL ANALYSIS")

ambiguous_words = [
    "diagnoses",
    "treated"
]

for word in ambiguous_words:

    print("\nWord:", word)

    analyses = parse_word(word)

    for analysis in analyses:

        stem, category, feature = analysis

        print(
            stem + "+" + category + "+" + feature
        )

    if len(analyses) > 1:
        print("Status: Ambiguous word")
    else:
        print("Status: Single analysis")


print("""
A downstream clinical NLP system can resolve ambiguity
using sentence context, grammatical information and
medical terminology.

For example, 'diagnoses' can represent a plural noun
or a third-person singular verb.
""")


# ----------------------------------------------
# 8. FINAL RESULT
# ----------------------------------------------

print("\nTASK 7 COMPLETED SUCCESSFULLY")
```

```python

# ==============================================
# TASK 8: ORTHOGRAPHIC RULES AND TWO-LEVEL MORPHOLOGY
# ==============================================

import re

# ----------------------------------------------
# 1. LEXICON-ONLY CONCATENATION PROBLEM
# ----------------------------------------------

print("1. LEXICON-ONLY CONCATENATION")

examples = [
    ("study", "s", "studys", "studies"),
    ("treat", "ing", "treating", "treating"),
    ("stop", "ed", "stoped", "stopped"),
    ("class", "s", "classs", "classes"),
    ("make", "ing", "makeing", "making")
]

for stem, suffix, wrong, correct in examples:

    print("\nStem:", stem)
    print("Suffix:", suffix)
    print("Direct Concatenation:", wrong)
    print("Correct Surface Form:", correct)


# ----------------------------------------------
# 2. ORTHOGRAPHIC REWRITE RULES
# ----------------------------------------------

print("\n2. ORTHOGRAPHIC REWRITE RULES")

rules = [
    "1. y -> i / consonant __ {s, ed}",
    "2. e -> 0 / __ ing (silent final e deletion)",
    "3. s -> es / {s, x, z, ch, sh} __ (sibilant plural)",
    "4. C -> CC / stressed CVC __ {ed, ing} (consonant doubling)",
    "5. 0 -> e / {s, d} __ after a final silent e (e retention)"
]

for rule in rules:
    print(rule)


# ----------------------------------------------
# 3. ORTHOGRAPHIC TRANSFORMATION FUNCTIONS
# ----------------------------------------------

def is_consonant_y(stem):

    return (
        len(stem) > 1
        and stem.endswith("y")
        and stem[-2] not in "aeiou"
    )


# Rule 1: Y-to-I conversion

def apply_y_rule(stem, suffix):

    if suffix in ("s", "ed") and is_consonant_y(stem):
        return stem[:-1] + "i"

    return stem


# Rule 2: Final E deletion before ING

def apply_e_deletion(stem, suffix):

    if suffix == "ing" and stem.endswith("e"):
        if stem.endswith("ee"):
            return stem

        return stem[:-1]

    return stem


# Rule 3: Sibilant -ES rule

def apply_sibilant_rule(stem, suffix):

    if suffix == "s" and stem.endswith(
        ("s", "x", "z", "ch", "sh")
    ):
        return stem + "e"

    return stem


# Rule 4: Consonant doubling

DOUBLING_WORDS = {
    "stop", "plan", "admit"
}

def apply_doubling_rule(stem, suffix):

    if suffix in ("ed", "ing"):

        if stem in DOUBLING_WORDS:
            return stem + stem[-1]

    return stem


# Rule 5: Final E retention

def apply_e_retention(stem, suffix):

    if stem.endswith("e") and suffix in ("s", "d"):
        return stem

    return stem


# ----------------------------------------------
# 4. RULE COMPOSITION
# ----------------------------------------------

def orthographic_transducer(stem, suffix):

    original_stem = stem

    # Apply relevant orthographic rules.

    if suffix == "ing":
        stem = apply_e_deletion(stem, suffix)

    elif suffix == "s":
        stem = apply_sibilant_rule(stem, suffix)

    elif suffix in ("s", "ed"):
        stem = apply_y_rule(stem, suffix)

    if suffix in ("ed", "ing"):
        stem = apply_doubling_rule(
            original_stem, suffix
        )

    # E retention takes priority over ordinary suffix addition.
    if original_stem.endswith("e") and suffix in ("s", "d"):
        stem = apply_e_retention(original_stem, suffix)

    return stem + suffix


# ----------------------------------------------
# 5. TWENTY TEST CASES
# ----------------------------------------------

print("\n3. ORTHOGRAPHIC RULE TESTING")

test_cases = [

    ("patient", "s", "patients"),
    ("fracture", "s", "fractures"),
    ("class", "s", "classes"),
    ("box", "s", "boxes"),
    ("baby", "s", "babies"),
    ("boy", "s", "boys"),

    ("study", "ed", "studied"),
    ("study", "ing", "studying"),

    ("treat", "ed", "treated"),
    ("treat", "ing", "treating"),

    ("make", "ing", "making"),
    ("see", "ing", "seeing"),

    ("stop", "ed", "stopped"),
    ("stop", "ing", "stopping"),
    ("plan", "ing", "planning"),
    ("admit", "ed", "admitted"),

    ("visit", "ed", "visited"),
    ("love", "d", "loved"),
    ("make", "s", "makes"),
    ("monitor", "ing", "monitoring")
]

success = 0
failure = 0

for stem, suffix, expected in test_cases:

    result = orthographic_transducer(stem, suffix)

    if result == expected:
        status = "PASS"
        success += 1
    else:
        status = "FAIL"
        failure += 1

    print(
        stem + "+" + suffix,
        "Expected:", expected,
        "Generated:", result,
        "Status:", status
    )


# ----------------------------------------------
# 6. TEST SUMMARY
# ----------------------------------------------

print("\n4. TEST SUMMARY")

total = len(test_cases)

print("Total Test Cases:", total)
print("Successful Cases:", success)
print("Failed Cases:", failure)

accuracy = (success / total) * 100

print("Success Percentage:", round(accuracy, 2), "%")


# ----------------------------------------------
# 7. CASCADE COMPOSITION
# ----------------------------------------------

print("\n5. FINITE-STATE CASCADE COMPOSITION")

print("""
Lexicon FST
     |
     v
Y-to-I Rule
     |
     v
E-Deletion Rule
     |
     v
Sibilant -ES Rule
     |
     v
Consonant Doubling Rule
     |
     v
E-Retention Rule
     |
     v
Final Surface Word
""")

print("Example:")
print("study + ed -> studied")
print("stop + ed -> stopped")
print("class + s -> classes")
print("make + ing -> making")


# ----------------------------------------------
# 8. TWO-LEVEL MORPHOLOGY
# ----------------------------------------------

print("\n6. TWO-LEVEL MORPHOLOGY")

print("""
Two-level morphology:

1. Lexical representation:
   stop + ed

2. Surface representation:
   stopped

The lexical and surface forms are related
through character-level correspondence rules.

Sequential Cascade:
- Rules are applied one after another.
- Rule ordering is important.

Parallel Two-Level Model:
- Rules operate simultaneously.
- Each lexical character is related
  to its corresponding surface character.

Advantages:
- Cascade: Easy to implement and debug.
- Two-level: Avoids repeated intermediate
  transformations and supports parallel constraints.
""")


# ----------------------------------------------
# 9. FAILURE ANALYSIS
# ----------------------------------------------

print("\n7. FAILURE ANALYSIS")

if failure == 0:
    print("All 20 test cases passed.")

else:
    print("Failed cases need additional spelling rules.")

    print("""
Possible causes:

1. Irregular spelling forms.
2. Exceptions to consonant doubling.
3. Morphological boundary conditions.
4. Stress-dependent spelling changes.
5. Lexical exceptions.
    """)


# ----------------------------------------------
# 10. FINAL RESULT
# ----------------------------------------------

print("\nTASK 8 COMPLETED SUCCESSFULLY")
```

```python

# ==============================================
# TASK 9: THE PORTER STEMMER
# ==============================================

import re
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# ----------------------------------------------
# 1. CONSONANT AND VOWEL DEFINITION
# ----------------------------------------------

print("1. CONSONANT AND VOWEL DEFINITION")

print("""
A vowel is a, e, i, o or u.
The letter y is treated as a consonant
at the beginning of a word and alternates
between consonant and vowel depending on context.

A measure m counts the number of
VC sequences in a word.

Example:
trouble = tr (VC sequence analysis) ...
""")

def is_consonant(word, i):

    ch = word[i]

    if ch in "aeiou":
        return False

    if ch == "y":
        if i == 0:
            return True
        return not is_consonant(word, i - 1)

    return True


def measure(word):

    # Count VC transitions.
    sequence = [
        "C" if is_consonant(word, i) else "V"
        for i in range(len(word))
    ]

    m = 0
    for i in range(1, len(sequence)):
        if sequence[i - 1] == "V" and sequence[i] == "C":
            m += 1

    return m


def has_vowel(word):

    return any(
        not is_consonant(word, i)
        for i in range(len(word))
    )


def ends_double_consonant(word):

    return (
        len(word) >= 2
        and word[-1] == word[-2]
        and is_consonant(word, len(word) - 1)
    )


def ends_cvc(word):

    if len(word) < 3:
        return False

    a = is_consonant(word, len(word) - 3)
    b = not is_consonant(word, len(word) - 2)
    c = is_consonant(word, len(word) - 1)

    return (
        a and b and c
        and word[-1] not in "wxy"
    )


# ----------------------------------------------
# 2. MEASURE CALCULATION
# ----------------------------------------------

print("\n2. MEASURE CALCULATION")

clinical_words = [
    "clinical",
    "fractures",
    "treatment",
    "infection",
    "monitoring",
    "diagnosis"
]

for word in clinical_words:
    print(word, "-> m =", measure(word))


# ----------------------------------------------
# 3. PORTER CONDITIONS
# ----------------------------------------------

print("\n3. PORTER CONDITIONS")

print("*v* : The stem contains a vowel.")
print("*d  : The stem ends with a double consonant.")
print("*o  : The stem ends in consonant-vowel-consonant,")
print("      excluding w, x and y as the final consonant.")


# ----------------------------------------------
# 4. FIVE PORTER STEPS
# ----------------------------------------------

print("\n4. FIVE PORTER STEPS")

def porter_trace(word):

    word = word.lower()

    print("\nOriginal Word:", word)

    current = word

    # Step 1a: Plurals
    current = stemmer._step1a(current)
    print("Step 1a:", current)

    # Step 1b: Past tense and -ing
    current = stemmer._step1b(current)
    print("Step 1b:", current)

    # Step 1c: y -> i
    current = stemmer._step1c(current)
    print("Step 1c:", current)

    # Step 2: Derivational suffixes
    current = stemmer._step2(current)
    print("Step 2:", current)

    # Step 3: Further suffix replacement
    current = stemmer._step3(current)
    print("Step 3:", current)

    # Step 4: Suffix removal
    current = stemmer._step4(current)
    print("Step 4:", current)

    # Step 5: Final e and l removal
    current = stemmer._step5a(current)
    current = stemmer._step5b(current)
    print("Step 5:", current)

    print("Final Stem:", stemmer.stem(word))


words = [
    "fractures",
    "treated",
    "monitoring",
    "infections",
    "clinical",
    "diagnoses"
]

for word in words:
    porter_trace(word)


# ----------------------------------------------
# 5. 200 ANNOTATED WORD FORMS
# ----------------------------------------------

print("\n5. STEMMING VS LEMMATIZATION")

# Illustrative 200-form benchmark:
# 20 lexical families with 10 forms each.

families = {
    "patient": [
        "patient", "patients", "patient's", "patiently",
        "patienthood", "patients'", "patientness",
        "patienting", "patiented", "patienter"
    ],
    "treat": [
        "treat", "treats", "treated", "treating",
        "treatable", "untreated", "retreat",
        "treatment", "treatments", "retreating"
    ],
    "monitor": [
        "monitor", "monitors", "monitored", "monitoring",
        "monitorable", "monitorings", "monitorer",
        "monitors'", "monitoredly", "remonitor"
    ],
    "fracture": [
        "fracture", "fractures", "fractured", "fracturing",
        "fracturable", "fractures'", "fracturement",
        "fracturer", "refracture", "fractureing"
    ],
    "infect": [
        "infect", "infects", "infected", "infecting",
        "infection", "infections", "infectious",
        "infective", "reinfection", "uninfected"
    ],
    "diagnos": [
        "diagnosis", "diagnoses", "diagnosed", "diagnosing",
        "diagnostic", "diagnostically", "rediagnosis",
        "undiagnosed", "diagnostics", "diagnosable"
    ],
    "cardi": [
        "cardiac", "cardiology", "cardiologist",
        "cardiologists", "cardiological", "cardiogram",
        "cardiograms", "cardiomyopathy",
        "cardiovascular", "carditis"
    ],
    "nephr": [
        "nephron", "nephrons", "nephritis",
        "nephritic", "nephrology", "nephrologist",
        "nephrologists", "nephrectomy",
        "nephropathic", "nephropathy"
    ],
    "gastr": [
        "gastric", "gastritis", "gastrectomy",
        "gastroenterology", "gastroenterologist",
        "gastralgia", "gastricly", "gastros",
        "gastropathic", "gastropathy"
    ],
    "severe": [
        "severe", "severer", "severest",
        "severely", "severity", "severities",
        "severeness", "severenesses",
        "severeest", "severenessly"
    ],
    "high": [
        "high", "higher", "highest", "highly",
        "highness", "highnesses", "highs",
        "highlyest", "highered", "highnessly"
    ],
    "class": [
        "class", "classes", "classed", "classing",
        "classification", "classifications",
        "classify", "classified", "classifiable",
        "classroom"
    ],
    "medic": [
        "medicine", "medicines", "medical",
        "medically", "medication", "medications",
        "medicate", "medicated", "medicating",
        "medicinal"
    ],
    "respir": [
        "respire", "respired", "respiring",
        "respiration", "respirations", "respiratory",
        "respirator", "respirators",
        "respiratoryly", "respiratoryness"
    ],
    "operat": [
        "operate", "operates", "operated", "operating",
        "operation", "operations", "operative",
        "operatively", "operator", "operators"
    ],
    "pain": [
        "pain", "pains", "pained", "paining",
        "painful", "painfully", "painless",
        "painlessly", "painfulness", "painkiller"
    ],
    "heal": [
        "heal", "heals", "healed", "healing",
        "healer", "healers", "health",
        "healthy", "healthier", "healthiest"
    ],
    "measure": [
        "measure", "measures", "measured", "measuring",
        "measurement", "measurements", "measurable",
        "measurably", "measurer", "remeasure"
    ],
    "examin": [
        "examine", "examines", "examined", "examining",
        "examination", "examinations", "examiner",
        "examiners", "examinable", "reexamine"
    ],
    "admit": [
        "admit", "admits", "admitted", "admitting",
        "admission", "admissions", "admissible",
        "admittedly", "readmit", "readmission"
    ]
}

annotated = []

for lemma, forms in families.items():
    for form in forms:
        annotated.append((form, lemma))

print("Total annotated forms:", len(annotated))

# This illustrative lemma map is manually supplied.
# It is not a general clinical morphological analyser.

def known_lemma(form):
    for lemma, forms in families.items():
        if form in forms:
            return lemma
    return form


agree = 0
diverge = 0

for form, lemma in annotated:

    porter_result = stemmer.stem(form)
    lemma_result = known_lemma(form)

    if porter_result == lemma_result:
        agree += 1
    else:
        diverge += 1

print("Agreement:", agree)
print("Divergence:", diverge)
print("Agreement Percentage:",
      round(agree / len(annotated) * 100, 2))


# ----------------------------------------------
# 6. OVER-STEMMING AND UNDER-STEMMING
# ----------------------------------------------

print("\n6. OVER-STEMMING AND UNDER-STEMMING")

distinct_pairs = [
    ("universe", "university"),
    ("general", "generate"),
    ("organ", "organization"),
    ("policy", "police")
]

related_pairs = [
    ("treat", "treated"),
    ("monitor", "monitoring"),
    ("fracture", "fractures"),
    ("infect", "infected")
]

over = 0

for a, b in distinct_pairs:
    if stemmer.stem(a) == stemmer.stem(b):
        over += 1
    print(a, b, "->", stemmer.stem(a), stemmer.stem(b))

under = 0

for a, b in related_pairs:
    if stemmer.stem(a) != stemmer.stem(b):
        under += 1
    print(a, b, "->", stemmer.stem(a), stemmer.stem(b))

print("Over-stemming:", over, "/", len(distinct_pairs))
print("Under-stemming:", under, "/", len(related_pairs))


# ----------------------------------------------
# 7. RETRIEVAL EXPERIMENT: 40 QUERIES
# ----------------------------------------------

print("\n7. RETRIEVAL EXPERIMENT")

documents = [
    "patient treated with medication",
    "patients received treatment",
    "fractures were monitored",
    "fracture monitoring was completed",
    "infection diagnosed in patient",
    "infections require treatment",
    "clinical examination completed",
    "cardiology reviewed the patient",
    "severe pain reported",
    "high blood pressure recorded"
]

queries = [
    "patient", "patients", "treat", "treated",
    "treatment", "fracture", "fractures", "monitor",
    "monitoring", "infection", "infections", "diagnosis",
    "diagnoses", "clinical", "examination", "cardiology",
    "cardiac", "pain", "severe", "high",
    "patients", "treating", "treatment", "fractured",
    "monitored", "infected", "infection", "diagnostic",
    "examined", "cardiologist", "painful", "higher",
    "treat", "fracture", "monitor", "infect",
    "diagnosis", "clinical", "patient", "severe"
]

assert len(queries) == 40


def tokens(text):
    return re.findall(r"[a-z]+", text.lower())


def retrieve(query, mode):

    results = []

    for index, document in enumerate(documents):

        doc_words = tokens(document)

        if mode == "none":
            matched = query in doc_words

        elif mode == "porter":
            matched = (
                stemmer.stem(query) in
                [stemmer.stem(w) for w in doc_words]
            )

        else:
            matched = (
                known_lemma(query) in
                [known_lemma(w) for w in doc_words]
            )

        if matched:
            results.append(index)

    return results


# Ground truth uses manually annotated lemma families.
def relevant_documents(query):

    lemma = known_lemma(query)
    relevant = []

    for index, document in enumerate(documents):

        if any(
            known_lemma(word) == lemma
            for word in tokens(document)
        ):
            relevant.append(index)

    return relevant


def evaluate(mode):

    total_tp = 0
    total_fp = 0
    total_fn = 0

    for query in queries:

        retrieved = set(retrieve(query, mode))
        relevant = set(relevant_documents(query))

        total_tp += len(retrieved & relevant)
        total_fp += len(retrieved - relevant)
        total_fn += len(relevant - retrieved)

    precision = (
        total_tp / (total_tp + total_fp)
        if total_tp + total_fp else 0
    )

    recall = (
        total_tp / (total_tp + total_fn)
        if total_tp + total_fn else 0
    )

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return precision, recall, f1


print("\nRetrieval Comparison")

for mode in ["none", "porter", "lemma"]:

    p, r, f = evaluate(mode)

    print(
        mode,
        "| Precision:", round(p, 3),
        "| Recall:", round(r, 3),
        "| F1:", round(f, 3)
    )


print("\nTASK 9 COMPLETED SUCCESSFULLY")
```

```python

# ==============================================
# TASK 10: IMPLEMENT AND TEST THE PIPELINE
# ==============================================

import re
import time
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# ----------------------------------------------
# 1. CLINICAL CORPUS
# ----------------------------------------------

corpus = [
    "Patient P101 is 45 years old.",
    "Blood pressure is 120/80 mmHg.",
    "Heart rate is 90 bpm.",
    "Temperature is 37.5 C.",
    "SpO2 is 98%.",
    "Patient has diabetes.",
    "Patient has hypertension.",
    "Prescribed 500 mg paracetamol.",
    "Take medicine twice daily.",
    "Treatment continued for 7 days.",
    "Patient received treatment.",
    "The patients have fractures.",
    "The doctor is monitoring infections.",
    "The diagnosis was confirmed.",
    "The diagnoses were recorded."
]


# ----------------------------------------------
# 2. TOKENIZER
# ----------------------------------------------

TOKEN_REGEX = re.compile(
    r"\d+(?:\.\d+)?(?:mg|mcg|ml|mmHg|bpm)\b"
    r"|\d{1,2}/\d{1,2}/\d{4}"
    r"|[A-Za-z]+(?:-[A-Za-z]+)*"
    r"|\d+(?:\.\d+)?"
    r"|[^\w\s]",
    re.I
)


def tokenize(text):
    return TOKEN_REGEX.findall(text)


# ----------------------------------------------
# 3. ENTITY PATTERNS
# ----------------------------------------------

patterns = {
    "Patient ID": r"\bP\d{3}\b",
    "Age": r"\b\d{1,3}\s*(?:years?|yrs?)\b",
    "Blood Pressure": r"\b\d{2,3}/\d{2,3}\s*mmHg\b",
    "Heart Rate": r"\b\d{2,3}\s*bpm\b",
    "Temperature": r"\b\d{2,3}(?:\.\d)?\s*°?C\b",
    "Oxygen Saturation": r"\bSpO2\s*[:=]?\s*\d{2,3}%?\b",
    "Dosage": r"\b\d+(?:\.\d+)?\s*(?:mg|mcg|g)\b",
    "Frequency": r"\b(?:once|twice|thrice|daily|bid|tid|qid)\b",
    "Duration": r"\b\d+\s*(?:days?|weeks?|months?)\b",
    "Diagnosis": r"\b(?:diabetes|hypertension|asthma|pneumonia)\b",
    "Medication": r"\b(?:paracetamol|metformin|ibuprofen|amoxicillin)\b",
    "Date": r"\b\d{2}/\d{2}/\d{4}\b"
}


# ----------------------------------------------
# 4. DFA VALIDATION
# ----------------------------------------------

def patient_dfa(value):

    transitions = {
        ("q0", "P"): "q1",
        ("q1", "D"): "q2",
        ("q2", "D"): "q3",
        ("q3", "D"): "q4"
    }

    state = "q0"

    for ch in value:
        symbol = "D" if ch.isdigit() else ch

        if (state, symbol) not in transitions:
            return False

        state = transitions[(state, symbol)]

    return state == "q4"


# ----------------------------------------------
# 5. NFA TO DFA SUBSET CONSTRUCTION
# ----------------------------------------------

nfa = {
    0: {"P": {1}},
    1: {"D": {2}},
    2: {"D": {3}},
    3: {"D": {4}}
}

alphabet = ["P", "D"]

start = frozenset({0})
states = [start]
queue = [start]
dfa_table = {}

while queue:

    current = queue.pop(0)

    for symbol in alphabet:

        target = set()

        for state in current:
            target.update(
                nfa.get(state, {}).get(symbol, set())
            )

        target = frozenset(target)
        dfa_table[(current, symbol)] = target

        if target not in states:
            states.append(target)
            queue.append(target)


print("SUBSET CONSTRUCTION")

for state in states:
    print(
        state,
        "P =", dfa_table.get((state, "P"), set()),
        "D =", dfa_table.get((state, "D"), set())
    )


# ----------------------------------------------
# 6. MORPHOLOGICAL TRANSDUCER
# ----------------------------------------------

lexicon = {
    "patients": ("patient", "N", "PL"),
    "fractures": ("fracture", "N", "PL"),
    "treated": ("treat", "V", "PAST"),
    "monitoring": ("monitor", "V", "GER"),
    "diagnoses": ("diagnosis", "N", "PL"),
    "diagnosed": ("diagnose", "V", "PAST"),
    "went": ("go", "V", "PAST"),
    "children": ("child", "N", "PL")
}


def morphological_parse(word):
    return lexicon.get(word.lower(), None)


# ----------------------------------------------
# 7. PORTER STEMMER
# ----------------------------------------------

def porter_analysis(text):

    return [
        stemmer.stem(word.lower())
        for word in tokenize(text)
        if word.isalpha()
    ]


# ----------------------------------------------
# 8. TEST CASES
# ----------------------------------------------

print("\nPIPELINE TESTING")

tests = [
    ("Patient ID", "P123", True),
    ("Patient ID", "P12", False),
    ("Age", "45 years", True),
    ("Age", "45", False),
    ("Blood Pressure", "120/80 mmHg", True),
    ("Blood Pressure", "120/80", False),
    ("Heart Rate", "90 bpm", True),
    ("Heart Rate", "90", False),
    ("Temperature", "37.5 C", True),
    ("Temperature", "37.5 F", False),
    ("Oxygen Saturation", "SpO2 98%", True),
    ("Oxygen Saturation", "98%", False),
    ("Dosage", "500 mg", True),
    ("Dosage", "500 ml", False),
    ("Frequency", "twice", True),
    ("Frequency", "sometimes", False),
    ("Duration", "7 days", True),
    ("Duration", "7", False),
    ("Diagnosis", "diabetes", True),
    ("Diagnosis", "headache", False),
    ("Medication", "paracetamol", True),
    ("Medication", "aspirin", False),
    ("Date", "06/11/2026", True),
    ("Date", "2026-11-06", False)
]

passed = 0
failed = 0

for entity, value, expected in tests:

    result = bool(
        re.fullmatch(patterns[entity], value, re.I)
    )

    status = result == expected

    if status:
        passed += 1
    else:
        failed += 1

    print(
        entity, value,
        "Expected:", expected,
        "Actual:", result,
        "PASS" if status else "FAIL"
    )


# ----------------------------------------------
# 9. COMPLETE PIPELINE AND RUNTIME
# ----------------------------------------------

print("\nCOMPLETE PIPELINE")

total_tokens = 0
total_entities = 0
total_analysed = 0

start_time = time.perf_counter()

for record in corpus:

    tokens = tokenize(record)
    total_tokens += len(tokens)

    for entity, pattern in patterns.items():
        matches = re.findall(pattern, record, re.I)
        total_entities += len(matches)

    for word in tokens:
        if word.isalpha():
            if morphological_parse(word):
                total_analysed += 1

    porter_analysis(record)

runtime = time.perf_counter() - start_time

print("Records Processed:", len(corpus))
print("Tokens Processed:", total_tokens)
print("Entities Matched:", total_entities)
print("Words Morphologically Analysed:", total_analysed)
print("Test Cases Passed:", passed)
print("Test Cases Failed:", failed)
print("Runtime:", round(runtime, 6), "seconds")


# ----------------------------------------------
# 10. PERFORMANCE TABLE
# ----------------------------------------------

print("\nPIPELINE PERFORMANCE TABLE")

stages = [
    ("Tokenization", len(corpus), total_tokens),
    ("Entity Recognition", len(corpus), total_entities),
    ("Morphological Analysis", total_tokens, total_analysed),
    ("Automata Testing", len(tests), passed)
]

print("Stage | Input | Matched | Rejected")

for stage, inputs, matched in stages:
    print(
        stage, "|", inputs, "|", matched,
        "|", inputs - matched
    )


# ----------------------------------------------
# 11. TEXT-BASED PERFORMANCE GRAPH
# ----------------------------------------------

print("\nPERFORMANCE GRAPH")

values = {
    "Tokenization": total_tokens,
    "Entities": total_entities,
    "Morphology": total_analysed
}

for name, value in values.items():

    bar = "#" * min(value, 50)

    print(name.ljust(20), bar, value)


# ----------------------------------------------
# 12. IRREGULAR WORD TESTING
# ----------------------------------------------

print("\nIRREGULAR WORD TESTING")

irregular_words = {
    "went": "go",
    "children": "child",
    "diagnoses": "diagnosis"
}

for word, expected in irregular_words.items():

    result = morphological_parse(word)

    print(
        word,
        "Expected:", expected,
        "Actual:", result[0] if result else "Not found"
    )


print("\nTASK 10 COMPLETED SUCCESSFULLY")
```

```python

# ==============================================
# TASK 11: COMPARE AND RECOMMEND NORMALISATION
# ==============================================

import re
import time
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# ----------------------------------------------
# 1. CLINICAL DOCUMENTS
# ----------------------------------------------

documents = [
    "patient received treatment",
    "patients were treated",
    "fractures were monitored",
    "fracture monitoring completed",
    "infection diagnosed",
    "infections require treatment",
    "clinical examination completed",
    "cardiology reviewed patient",
    "severe pain reported",
    "high blood pressure recorded"
]

queries = [
    "patient", "patients", "treat", "treated",
    "treatment", "fracture", "fractures", "monitor",
    "monitoring", "infection", "infections", "diagnosis",
    "diagnoses", "clinical", "examination", "cardiology",
    "cardiac", "pain", "severe", "high"
]

# ----------------------------------------------
# 2. MORPHOLOGICAL LEXICON
# ----------------------------------------------

lexicon = {
    "patients": "patient",
    "treated": "treat",
    "treating": "treat",
    "treatment": "treat",
    "fractures": "fracture",
    "monitored": "monitor",
    "monitoring": "monitor",
    "infections": "infection",
    "diagnoses": "diagnosis",
    "diagnostic": "diagnosis",
    "examined": "examine",
    "examination": "examine"
}


def words(text):
    return re.findall(r"[a-z]+", text.lower())


def lemma(word):
    return lexicon.get(word, word)


# ----------------------------------------------
# 3. RETRIEVAL METHODS
# ----------------------------------------------

def retrieve(query, mode):

    results = []

    for index, document in enumerate(documents):

        doc_words = words(document)

        if mode == "No Normalisation":
            match = query in doc_words

        elif mode == "Porter Stemming":
            match = (
                stemmer.stem(query) in
                [stemmer.stem(w) for w in doc_words]
            )

        else:
            match = (
                lemma(query) in
                [lemma(w) for w in doc_words]
            )

        if match:
            results.append(index)

    return results


# ----------------------------------------------
# 4. GROUND-TRUTH RELEVANCE
# ----------------------------------------------

def relevant_documents(query):

    target = lemma(query)
    result = []

    for i, document in enumerate(documents):

        if any(
            lemma(word) == target
            for word in words(document)
        ):
            result.append(i)

    return result


# ----------------------------------------------
# 5. PRECISION, RECALL AND F1
# ----------------------------------------------

def evaluate(mode):

    tp = fp = fn = 0

    for query in queries:

        retrieved = set(retrieve(query, mode))
        relevant = set(relevant_documents(query))

        tp += len(retrieved & relevant)
        fp += len(retrieved - relevant)
        fn += len(relevant - retrieved)

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return precision, recall, f1


# ----------------------------------------------
# 6. RUNTIME MEASUREMENT
# ----------------------------------------------

print("1. NORMALISATION COMPARISON")

methods = [
    "No Normalisation",
    "Porter Stemming",
    "Finite-State Analysis"
]

results = {}

for method in methods:

    start = time.perf_counter()

    for query in queries:
        retrieve(query, method)

    elapsed = time.perf_counter() - start

    p, r, f = evaluate(method)

    results[method] = {
        "precision": p,
        "recall": r,
        "f1": f,
        "runtime": elapsed
    }


# ----------------------------------------------
# 7. COMPARISON TABLE
# ----------------------------------------------

print("\n2. PERFORMANCE COMPARISON")

print(
    "Method | Precision | Recall | F1 | Runtime"
)

for method, data in results.items():

    print(
        method,
        "|", round(data["precision"], 3),
        "|", round(data["recall"], 3),
        "|", round(data["f1"], 3),
        "|", round(data["runtime"], 6)
    )


# ----------------------------------------------
# 8. COVERAGE COMPARISON
# ----------------------------------------------

print("\n3. ANALYSER COVERAGE")

vocabulary = set(
    word
    for document in documents
    for word in words(document)
)

known_forms = set(lexicon) | set(lexicon.values())

coverage = (
    len(vocabulary & known_forms) / len(vocabulary) * 100
    if vocabulary else 0
)

print("Total Unique Words:", len(vocabulary))
print("Known Morphological Forms:",
      len(vocabulary & known_forms))
print("Lexicon Coverage:", round(coverage, 2), "%")


# ----------------------------------------------
# 9. MAINTENANCE EFFORT
# ----------------------------------------------

print("\n4. MAINTENANCE EFFORT")

effort = {
    "No Normalisation": "Low",
    "Porter Stemming": "Low",
    "Finite-State Analysis": "High"
}

for method, level in effort.items():
    print(method, ":", level)


# ----------------------------------------------
# 10. FINAL RECOMMENDATION
# ----------------------------------------------

print("\n5. RECOMMENDATION")

print("""
Recommended Design:
Finite-State Morphological Analysis with
Porter stemming as a controlled fallback.

Reasons:
1. Preserves grammatical information.
2. Handles known clinical word forms.
3. Supports irregular word mappings.
4. Reduces incorrect conflation.
5. Allows unknown words to be handled separately.

Limitations:
1. Requires lexicon maintenance.
2. New clinical terms need annotation.
3. Ambiguous words require context.
4. Incorrect mappings may affect retrieval.

The final choice should consider precision,
recall, F1-score, runtime and clinical safety.
""")

print("\nTASK 11 COMPLETED SUCCESSFULLY")
```

```python
import re
import time

records = {

    "OCR": (
        "P1O1: 5O0mg paracetamol prescribed "
        "for hypertenslon."
    ),

    "CODE_MIXED": (
        "patient ku fever irundhuchu, "
        "paracetamol 500mg kuduthen."
    ),

    "COMBINED": (
        "P1O2: hypertenslon irundhuchu. "
        "5O0mg paracetamol kuduthen."
    )
}

print("1. NOISY CLINICAL RECORDS")

for name, record in records.items():
    print(name, ":", record)



original_patterns = {
    "Patient ID": r"\bP\d{3}\b",
    "Dosage": r"\b\d+(?:\.\d+)?\s*(?:mg|mcg|g)\b",
    "Diagnosis": r"\b(?:diabetes|hypertension|asthma|pneumonia)\b",
    "Medication": r"\b(?:paracetamol|metformin|ibuprofen)\b",
    "Temperature": r"\b\d{2,3}(?:\.\d)?\s*°?C\b"
}


def original_pipeline(text):

    result = {}

    for entity, pattern in original_patterns.items():

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        result[entity] = matches

    return result



def correct_ocr(text):

    text = re.sub(
        r"-\s*\n\s*",
        "",
        text
    )
    text = re.sub(
        r"(?<=\d)O(?=\d)",
        "0",
        text
    )

    text = re.sub(
        r"(?<=\d)l(?=\d)",
        "1",
        text
    )

    # Contextual correction for common OCR confusion.
    text = re.sub(
        r"\bP1O(\d)\b",
        r"P10\1",
        text
    )

    return text



def edit_distance(a, b):

    rows = len(a) + 1
    cols = len(b) + 1

    dp = [
        [0] * cols
        for _ in range(rows)
    ]

    for i in range(rows):
        dp[i][0] = i

    for j in range(cols):
        dp[0][j] = j

    for i in range(1, rows):
        for j in range(1, cols):

            cost = 0 if a[i-1] == b[j-1] else 1

            dp[i][j] = min(
                dp[i-1][j] + 1,
                dp[i][j-1] + 1,
                dp[i-1][j-1] + cost
            )

    return dp[-1][-1]



clinical_vocabulary = [
    "hypertension",
    "diabetes",
    "asthma",
    "pneumonia",
    "paracetamol",
    "metformin",
    "ibuprofen",
    "fever",
    "patient"
]


def correct_spelling(word):

    word = word.lower()

    if word in clinical_vocabulary:
        return word

    candidates = [
        (edit_distance(word, term), term)
        for term in clinical_vocabulary
    ]

    distance, best_word = min(candidates)
.
    if distance <= 2:
        return best_word

    return word



tamil_dictionary = {
    "patient": "patient",
    "ku": "to",
    "fever": "fever",
    "irundhuchu": "had",
    "paracetamol": "paracetamol",
    "kuduthen": "administered"
}


def normalize_code_mixed(text):

    words = re.findall(
        r"[A-Za-z]+|\d+(?:\.\d+)?|[^\w\s]",
        text.lower()
    )

    output = []

    for word in words:

        output.append(
            tamil_dictionary.get(word, word)
        )

    return " ".join(output)



tamil_tokens = {
    "ku",
    "irundhuchu",
    "kuduthen",
    "vandhu",
    "saapida",
    "illai"
}


def identify_language(token):

    if token.lower() in tamil_tokens:
        return "Romanised Tamil"

    if re.fullmatch(r"\d+(?:\.\d+)?", token):
        return "Numeric"

    if token.lower() in clinical_vocabulary:
        return "Clinical English"

    return "Unknown / English"



def modified_pipeline(text):

    text = correct_ocr(text)

    text = normalize_code_mixed(text)

    corrected = []

    for word in text.split():

        cleaned = re.sub(r"[^A-Za-z0-9.]", "", word)

        if cleaned:
            fixed = correct_spelling(cleaned)

            corrected.append(
                word.replace(cleaned, fixed)
            )
        else:
            corrected.append(word)

    text = " ".join(corrected)

    entities = original_pipeline(text)

    return text, entities



print("\n2. ORIGINAL VS MODIFIED PIPELINE")

for name, record in records.items():

    print("\nRECORD:", name)

    original = original_pipeline(record)

    normalized, modified = modified_pipeline(record)

    print("\nOriginal Record:")
    print(record)

    print("\nOriginal Pipeline Results:")
    print(original)

    print("\nModified Record:")
    print(normalized)

    print("\nModified Pipeline Results:")
    print(modified)


print("\n3. TOKEN-LEVEL LANGUAGE IDENTIFICATION")

sample = records["CODE_MIXED"]

for token in re.findall(r"[A-Za-z]+", sample.lower()):

    print(
        token,
        "->",
        identify_language(token)
    )


print("\n4. NOISY DATA EVALUATION")

ground_truth = {
    "OCR": {
        "Patient ID": {"P101"},
        "Dosage": {"500mg"},
        "Diagnosis": {"hypertension"},
        "Medication": {"paracetamol"}
    },

    "CODE_MIXED": {
        "Dosage": {"500mg"},
        "Medication": {"paracetamol"}
    },

    "COMBINED": {
        "Patient ID": {"P102"},
        "Dosage": {"500mg"},
        "Diagnosis": {"hypertension"},
        "Medication": {"paracetamol"}
    }
}


def extract_entities(text):

    result = set()

    for entity, pattern in original_patterns.items():

        for match in re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        ):
            result.add((entity, match.lower()))

    return result


def expected_entities(labels):

    result = set()

    for entity, values in labels.items():

        for value in values:
            result.add((entity, value.lower()))

    return result


def calculate_metrics(predicted, expected):

    tp = len(predicted & expected)
    fp = len(predicted - expected)
    fn = len(expected - predicted)

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return tp, fp, fn, precision, recall, f1


for name, record in records.items():

    expected = expected_entities(
        ground_truth[name]
    )

    original_predicted = extract_entities(record)

    normalized, modified_entities = modified_pipeline(record)

    modified_predicted = set()

    for entity, values in modified_entities.items():

        for value in values:
            modified_predicted.add(
                (entity, value.lower())
            )

    original_metrics = calculate_metrics(
        original_predicted, expected
    )

    modified_metrics = calculate_metrics(
        modified_predicted, expected
    )

    print("\nRecord:", name)

    print("Original TP, FP, FN:",
          original_metrics[:3])

    print("Original Precision, Recall, F1:",
          [round(x, 3) for x in original_metrics[3:]])

    print("Modified TP, FP, FN:",
          modified_metrics[:3])

    print("Modified Precision, Recall, F1:",
          [round(x, 3) for x in modified_metrics[3:]])



print("\n5. RUNTIME COMPARISON")

for name, record in records.items():

    start = time.perf_counter()
    original_pipeline(record)
    original_time = time.perf_counter() - start

    start = time.perf_counter()
    modified_pipeline(record)
    modified_time = time.perf_counter() - start

    print("\nRecord:", name)
    print("Original Runtime:", round(original_time, 6))
    print("Modified Runtime:", round(modified_time, 6))


print("\nTASK 12 COMPLETED SUCCESSFULLY")
```
