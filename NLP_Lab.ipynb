```python
import re
from collections import Counter

text = """Python is a programming language.
Python is easy to learn.
Programming is useful."""

sentences = re.split(r'[.!?]+', text)
sentences = [s.strip() for s in sentences if s.strip()]

print("Sentences:")
for sentence in sentences:
    print(sentence)

words = re.findall(r'\b\w+\b', text.lower())

print("\nWords:")
print(words)

print("\nTotal number of words:", len(words))

word_count = Counter(words)
most_common_word, frequency = word_count.most_common(1)[0]

print("Most frequent word:", most_common_word)
print("Frequency:", frequency)
```

```python
from collections import Counter

text = """
the man went to the park
the man read a book in the park
the boy went to the park
the man likes the book
the book is good
"""
words = text.lower().split()
word_frequency = Counter(words)

print("Word Frequency Model:")
print(word_frequency)

def edit_distance(word1, word2):

    rows = len(word1) + 1
    cols = len(word2) + 1

    matrix = [[0] * cols for _ in range(rows)]

    for i in range(rows):
        matrix[i][0] = i

    for j in range(cols):
        matrix[0][j] = j

    for i in range(1, rows):
        for j in range(1, cols):

            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1

            matrix[i][j] = min(
                matrix[i - 1][j] + 1,
                matrix[i][j - 1] + 1,
                matrix[i - 1][j - 1] + cost
            )

    return matrix[-1][-1]


def spell_check(word):

    candidates = []

    for dictionary_word in word_frequency:

        distance = edit_distance(word, dictionary_word)

        if distance <= 1:
            candidates.append(dictionary_word)

    if not candidates:
        return "No correction found"

    best_word = max(
        candidates,
        key=lambda x: word_frequency[x]
    )

    return best_word

test_words = ["teh", "manni", "boook", "parkt"]

print("\nSpell Check Results:")

for word in test_words:
    correction = spell_check(word)
    print(word, "->", correction)
```

```python
import re

text = """Call John at 9876543210 or email john@gmail.com.
Meeting date is 25-10-2026.
Contact Mary at 9123456780.
Another date is 05-11-2026."""

print("Original Text:")
print(text)

first_date = re.search(r'\d{2}-\d{2}-\d{4}', text)

print("\nFirst Date:")
print(first_date.group())


mobile_numbers = re.findall(r'\b[6-9]\d{9}\b', text)

print("\nMobile Numbers:")
print(mobile_numbers)


emails = re.findall(
    r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',
    text
)

print("\nEmail IDs:")
print(emails)

dates = re.findall(r'\d{2}-\d{2}-\d{4}', text)

print("\nAll Dates:")
print(dates)


capital_words = re.findall(r'\b[A-Z][a-z]+\b', text)

print("\nCapitalized Words:")
print(capital_words)

if re.match(r'^Call', text):
    print("\nText starts with 'Call'")
else:
    print("\nText does not start with 'Call'")

new_text = re.sub(
    r'(\d{2})-(\d{2})-(\d{4})',
    r'\3/\2/\1',
    text
)

print("\nAfter changing date format:")
print(new_text)

hidden_text = re.sub(
    r'\b[6-9]\d{9}\b',
    'Call',
    text
)

print("\nAfter hiding phone numbers:")
print(hidden_text)
```

```python
def fsa(input_string):

    state = "q0"

    path = [state]

    for ch in input_string:

        if state == "q0":

            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        elif state == "q1":

            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q2"

        elif state == "q2":

            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q0"

        path.append(state)

    print("Input:", input_string)
    print("State Path:", " -> ".join(path))

    if state == "q2":
        print("Result: ACCEPTED")
    else:
        print("Result: REJECTED")

    print()

strings = ["ab", "aab", "bab", "aba", "bbb", "aaab"]

for s in strings:
    fsa(s)
```

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "running",
    "books",
    "cars"
]

print("MORPHOLOGICAL ANALYSIS")
print("-" * 30)

for word in words:

    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)

    print("Word:", word)
    print("Stem:", stem)
    print("Lemma:", lemma)
    print()
```

```python
def plural_fsm(word):

    if word.endswith(("s", "x", "z", "ch", "sh")):

        state = "q1"
        plural = word + "es"

    elif word.endswith("y") and len(word) > 1 \
            and word[-2] not in "aeiou":

        state = "q2"
        plural = word[:-1] + "ies"

    elif word.endswith("f"):

        state = "q3"
        plural = word[:-1] + "ves"

    else:

        state = "q4"
        plural = word + "s"

    return state, plural

words = [
    "cat",
    "book",
    "bus",
    "box",
    "church",
    "dish",
    "baby",
    "city",
    "leaf"
]

print("FINITE-STATE MORPHOLOGICAL PARSER")
print("-" * 40)

for word in words:

    state, plural = plural_fsm(word)

    print("Word:", word)
    print("State:", state)
    print("Plural:", plural)
    print()
```

```python
import re
resume = """
Name: swetha
Email: swethaselvam2809@gmail.com
Phone: 9876543210
Skills: Python, Java, SQL, Machine Learning, NLP
Experience: 3 years
"""
name = re.search(r'Name:\s*([A-Za-z ]+)', resume)

email = re.search(
    r'[\w.-]+@[\w.-]+\.\w+',
    resume
)

mobile = re.search(r'\b[6-9]\d{9}\b', resume)

skills_list = ["Python", "Java", "SQL", "Machine Learning", "NLP"]

found_skills = []

for skill in skills_list:
    if re.search(r'\b' + re.escape(skill) + r'\b', resume, re.IGNORECASE):
        found_skills.append(skill)

experience = re.search(r'Experience:\s*(\d+)\s*years?', resume)

print("RESUME INFORMATION")
print("-" * 30)

print("Candidate Name:",
      name.group(1).strip() if name else "Not Found")

print("Email:",
      email.group() if email else "Not Found")

print("Mobile Number:",
      mobile.group() if mobile else "Not Found")

print("Technical Skills:", ", ".join(found_skills))

years = int(experience.group(1)) if experience else 0
print("Experience:", years, "years")

if years >= 2 and "Python" in found_skills:
    print("\nEligibility: ELIGIBLE")
else:
    print("\nEligibility: NOT ELIGIBLE")

print("\nSTRUCTURED SUMMARY")
print("-" * 30)
print("Name       :", name.group(1).strip())
print("Email      :", email.group())
print("Mobile     :", mobile.group())
print("Skills     :", ", ".join(found_skills))
print("Experience :", years, "years")
```

```python

import re
products = [
    "Apple iPhone 15",
    "Apple iPhone 14",
    "Samsung Galaxy S24",
    "Samsung Galaxy A55",
    "OnePlus 12",
    "OnePlus Nord CE",
    "HP Pavilion Laptop",
    "Dell Inspiron Laptop",
    "Apple MacBook Air",
    "Sony Wireless Headphones"
]

def search_products(pattern, description):
    matches = []

    for product in products:
        if re.search(pattern, product, re.IGNORECASE):
            matches.append(product)

    print("\n" + description)
    print("-" * 40)

    if matches:
        for product in matches:
            print(product)
    else:
        print("No products found")

    print("Total matches:", len(matches))

keyword = "OnePlus"

pattern = r'^OnePlus 12$'
search_products(pattern, "1. Exact Keyword Search")

prefix = "Apple"

pattern = r'^' + re.escape(prefix)
search_products(pattern, "2. Prefix Search - Apple")

suffix = "Laptop"

pattern = re.escape(suffix) + r'$'
search_products(pattern, "3. Suffix Search - Laptop")

keyword = "Galaxy"

pattern = re.escape(keyword)
search_products(pattern, "4. Partial Keyword Search - Galaxy")

keyword = "apple"

pattern = re.escape(keyword)
search_products(pattern, "5. Case-Insensitive Search - apple")
```

```python
import re
student = {
    "Register Number": "24CSE001",
    "Email": "student@gmail.com",
    "Course Code": "CSE101",
    "Semester": "5",
    "Mobile Number": "9876543210"
}

patterns = {
    "Register Number": r'^\d{2}[A-Z]{3}\d{3}$',
    "Email": r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$',
    "Course Code": r'^[A-Z]{3}\d{3}$',
    "Semester": r'^[1-8]$',
    "Mobile Number": r'^[6-9]\d{9}$'
}

print("UNIVERSITY REGISTRATION VALIDATION")
print("-" * 40)

all_valid = True

for field, pattern in patterns.items():

    value = student[field]

    if re.fullmatch(pattern, value):
        print(field + ":", value, "-> VALID")
    else:
        print(field + ":", value, "-> INVALID")
        all_valid = False

print("\nFINAL REGISTRATION STATUS")
print("-" * 40)

if all_valid:
    print("Registration Successful")
else:
    print("Registration Failed")
```
